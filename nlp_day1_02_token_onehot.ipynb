## 1. 이번 실습에서 만들 것

하나의 문장을 **텍스트 → 토큰 → ID → One-hot 벡터**로 바꾸고, 각 숫자의 의미와 행렬의 크기(Shape)를 확인합니다.

- **A형 함께 구현하기:** `나는 사과를 좋아한다` → ID `[0, 1, 2]` → One-hot Shape `(3, 5)`
- **B형 직접 완성하기:** `나는 바나나를 먹는다` → `tokens_to_ids()` 완성 → ID `[0, 3, 4]`

필요한 라이브러리: Python, NumPy. 어휘집과 변환 함수는 두 예제에서 동일하게 사용합니다.

## 2. A형 · 함께 구현하기

### 2.1 텍스트 전처리와 토큰화

먼저 불필요한 공백을 정리합니다. 이번 예제에서는 이해하기 쉬운 **띄어쓰기 기준 토큰화**를 사용합니다.

```python
import re
import numpy as np

text_a = "  나는   사과를   좋아한다  "

# [텍스트 전처리]
# re.sub(): 연속된 공백을 하나의 공백으로 통일
# strip(): 문장 앞뒤의 불필요한 공백 제거
# 여러 공백을 하나로 합치고, 앞뒤 공백을 제거합니다.
clean_a = re.sub(r"\s+", " ", text_a).strip()
# [토큰화]
# split(): 정리된 문장을 띄어쓰기 기준으로 분리
tokens_a = clean_a.split()

print("정리된 문장:", clean_a)
print("토큰 목록:", tokens_a)
print("토큰 수:", len(tokens_a))

assert tokens_a == ["나는", "사과를", "좋아한다"]
```

### 2.2 어휘집과 토큰 ID

어휘집은 **토큰과 고유 ID를 연결하는 기준**입니다. 숫자의 크기는 단어의 의미나 중요도를 뜻하지 않습니다.

```python
# [어휘집 생성]
# 각 토큰에 고유한 정수 ID를 부여
# 같은 어휘집을 사용하면 같은 토큰은 항상 같은 ID로 변환됨
vocab = {
    "나는": 0,
    "사과를": 1,
    "좋아한다": 2,
    "바나나를": 3,
    "먹는다": 4,
}

# [토큰 ID 변환]
# tokens_a의 각 토큰을 vocab에서 찾아서 ID로 변환
# 나는 -> 0, 사과를 -> 1, 좋아한다 -> 2
ids_a = [vocab[token] for token in tokens_a]

print("어휘집 크기:", len(vocab))
print("A형 토큰 ID:", ids_a)

# 예산한 ID 목록과 일치하는지 검사
assert ids_a == [0, 1, 2]
```

### 2.3 One-hot 벡터 만들기

어휘집의 크기가 5이므로 각 토큰은 길이 5인 벡터가 됩니다. 각 행에서 **토큰 ID에 해당하는 열만 1**입니다.

```python
# [one-hot 인코딩]
# token_ids: 토큰 ID 목록, vocab_size: 어휘집 크기 (one-hot 벡터의 길이)
def ids_to_one_hot(token_ids, vocab_size):
    # 먼저 모든 값을 0으로 채운 행렬을 만듦
    one_hot = np.zeros((len(token_ids), vocab_size), dtype=np.int64)
    # row: 문장 안에서 몇번째 토큰인지 나타내는 행번호, token_id: 해당 토큰이 어휘집에서 차지하는 열번호
    for row, token_id in enumerate(token_ids):
        # 해당 토큰 ID에 해당하는 위치에만 1로 바꿈
        one_hot[row, token_id] = 1
    return one_hot

# A형 토큰 ID를 one-hot 인코딩으로 변환
one_hot_a = ids_to_one_hot(ids_a, len(vocab))
print("A형 One-hot 행렬:")
print(one_hot_a)
# 행 = 문장의 토큰수, 열 = 어휘집의 크기
print("Shape:", one_hot_a.shape)

assert one_hot_a.shape == (3, 5)
assert np.all(one_hot_a.sum(axis=1) == 1)
```

**결과 해석:** 행은 문장의 토큰 3개이고, 열은 어휘집에 등록된 토큰 5개입니다. 각 행에서 1의 위치가 토큰 ID와 일치하는지 확인해 봅시다.
